## 📚 Conceitos Fundamentais - Entendendo os Dados

Antes de começar a análise, é importante entender os **termos específicos** que usamos neste projeto:

### **1️⃣ TELEMETRIA** 
Dados coletados automaticamente pelos **sensores dos equipamentos** em tempo real. Exemplo:
- 📊 Temperatura de funcionamento
- 🔊 Nível de vibração
- 💨 Pressão interna
- ⚡ Consumo de energia

**Como funciona:** Cada equipamento envia milhões de leituras por mês (nossos dados: **37 milhões de registros em 6 meses**).

---

### **2️⃣ APONTAMENTOS**
**Eventos de parada ou anomalia** registrados manualmente pelos operadores. Exemplo:
- ❌ "Equipamento parou por falha no motor"
- ⚠️ "Vibração anormal detectada"
- 🔧 "Manutenção preventiva realizada"

**Nossa base:** 377.907 apontamentos em 6 meses

---

### **3️⃣ ALERTAS e CRITICIDADE**
Cada leitura de telemetria gera um **alerta categorizável**:

| Nível | Nome | Significado | Ação |
|-------|------|-------------|------|
| 🟢 **Verde** | Informacional | Sistema funcionando normalmente | Monitorar |
| 🟡 **Amarelo** | Não Crítico | Anomalia detectada, mas aceitável | Observar |
| 🔴 **Vermelho** | **Crítico (Don't Go)** | ⚠️ EQUIPAMENTO NÃO PODE OPERAR | Ação Imediata! |

**"Don't Go"** = Equipamento deve parar IMEDIATAMENTE para evitar danos maiores.

---

### **4️⃣ CORRELAÇÃO**
Mede se **duas variáveis andam juntas**. Exemplo:
- ✅ Temperatura alta + Vibração alta = Forte correlação positiva
- ❌ Temperatura alta + Pressão baixa = Correlação fraca

**Escala:** -1 (opostos) até +1 (iguais)

---

### **5️⃣ ROC-AUC (Métrica de Qualidade)**
Score que mede se nosso modelo de **previsão é bom**:
- 🟢 **0.8 a 1.0:** ✅ Excelente - confiável
- 🟡 **0.7 a 0.8:** ✓ Bom - aceitável  
- 🔴 **< 0.7:** ❌ Ruim - precisa melhorar

**Nossa meta:** ROC-AUC > 0.80

### O que é EDA?

**Análise Exploratória de Dados (EDA)** é a fase crucial onde:
- ✅ Descobrimos **padrões** nos dados
- ✅ Identificamos **correlações** entre variáveis
- ✅ Encontramos **anomalias e outliers**
- ✅ Preparamos **features** para modelos preditivos

````python
# =======================================================================================
# PREPARAÇÃO: IDENTIFICAÇÃO DE COLUNAS NUMÉRICAS
# =======================================================================================
# Extrai apenas as colunas numéricas do dataset de telemetria para análise de correlação
# Exclui colunas de data/hora e identificadores
# =======================================================================================

print("🔍 Identificando colunas numéricas para análise de correlação...\n")

# Selecionar apenas colunas numéricas (excluir object, datetime, etc)
num_cols = df_telemetria_total.select_dtypes(include=['number']).columns.tolist()

# Remover coluna de target se houver
if 'eh_critico' in num_cols:
    num_cols.remove('eh_critico')

print(f"✓ Total de colunas numéricas: {len(num_cols)}")
print(f"\nColunas identificadas:")
for col in num_cols[:10]:  # Mostrar primeiras 10
    print(f"  - {col}")
if len(num_cols) > 10:
    print(f"  ... e mais {len(num_cols) - 10} colunas")
````

````python
# =======================================================================================
# 🧠 ANÁLISE DE CORRELAÇÃO - MATRIZ DE CORRELAÇÃO 🧠
# =======================================================================================
#
# 👶 EXPLICAÇÃO PARA ENTENDER COMO SE FOSSE UMA CRIANÇA:
#
# CORRELAÇÃO = "Duas coisas que ANDAM JUNTAS"
#
# EXEMPLO 1: TEMPERATURA E SORVETE
# ────────────────────────────────────────────────────────
# Dia quente (temperatura SOBE) → Mais gente COMPRA sorvete
# Dia frio  (temperatura DESCE) → Menos gente COMPRA sorvete
#
# Resultado: CORRELAÇÃO POSITIVA!
# (Quando uma sobe, a outra também sobe)
#
# ────────────────────────────────────────────────────────
# EXEMPLO 2: EXERCÍCIO E CANSAÇO
# ────────────────────────────────────────────────────────
# Quando você EXERCITA MUITO → CANSAÇO é ALTO
# Quando você DESCANSA MUITO → CANSAÇO é BAIXO
#
# Resultado: CORRELAÇÃO POSITIVA!
# (Aumenta exercício = Aumenta cansaço)
#
# ────────────────────────────────────────────────────────
# EXEMPLO 3: VELOCIDADE DO CARRO E TEMPO DE VIAGEM
# ────────────────────────────────────────────────────────
# Você DIRIGE RÁPIDO (100 km/h) → Chega LOGO (tempo CURTO)
# Você DIRIGE LENTO (50 km/h)   → Chega DEPOIS (tempo LONGO)
#
# Resultado: CORRELAÇÃO NEGATIVA!
# (Aumenta velocidade = Diminui tempo)
#
# ────────────────────────────────────────────────────────
# ESCALA: De -1 até +1
# ────────────────────────────────────────────────────────
# +1.0 = PERFEITO POSITIVO (andam PERFEITAMENTE juntas)
# +0.8 = FORTE POSITIVO (andam muito juntas)
# +0.5 = MODERADO POSITIVO (andam um pouco juntas)
#  0.0 = NENHUMA RELAÇÃO (andam completamente separadas)
# -0.5 = MODERADO NEGATIVO (andam um pouco opostas)
# -0.8 = FORTE NEGATIVO (andam muito opostas)
# -1.0 = PERFEITO NEGATIVO (andam PERFEITAMENTE opostas)
#
# ────────────────────────────────────────────────────────
# AS CORES NO GRÁFICO:
# ────────────────────────────────────────────────────────
# 🔴 VERMELHO QUENTE    = Correlação POSITIVA forte
# ⚪ BRANCO             = Sem correlação
# 🔵 AZUL GELADO        = Correlação NEGATIVA forte
#
# =======================================================================================

# PASSO 1: Calcular correlação entre todas as variáveis numéricas
if len(num_cols) > 1:
    print("\n" + "="*80)
    print("📊 CALCULANDO MATRIZ DE CORRELAÇÃO")
    print("="*80)
    print("\n🤔 O que estou fazendo?")
    print("   Procurando por 'amizades' entre os números...")
    print("   Será que temperatura ANDA JUNTA com alerta crítico?")
    print("   Será que vibração ANDA JUNTA com pressão?\n")
    
    # Calcular a matriz
    correlation_matrix = df_telemetria_total[num_cols].corr()
    
    # Criar o heatmap (mapa de calor)
    plt.figure(figsize=(12, 8))
    sns.heatmap(
        correlation_matrix, 
        annot=True,  # Mostra o número em cada quadradinho
        cmap='coolwarm',  # Azul (negativo) até Vermelho (positivo)
        center=0,  # Cores iguais em zero
        fmt='.2f',  # Formato com 2 decimais
        square=True,  # Quadrados iguais
        cbar_kws={'label': 'Nível de Correlação (de -1 até +1)'},
        linewidths=0.5,
        linecolor='gray'
    )
    
    plt.title('🔗 MATRIZ DE CORRELAÇÃO - VARIÁVEIS NUMÉRICAS\n' + 
              '(Vermelho = Andam Juntas | Azul = Andam Opostas | Branco = Independentes)', 
              fontsize=14, fontweight='bold', color='#007E7A', pad=20)
    plt.tight_layout()
    plt.show()
    
    # PASSO 2: Interpretação detalhada
    print("\n" + "="*80)
    print("🔍 COMO LER ESTE GRÁFICO?")
    print("="*80)
    print("\n1️⃣ PROCURE PELA DIAGONAL (de canto a canto)")
    print("   └─→ Sempre será VERMELHA (correlação = 1.0)")
    print("   └─→ Por quê? Porque uma coisa com ela mesma é PERFEITA!")
    print("   └─→ (Temperatura com temperatura = 100% igual)")
    
    print("\n2️⃣ PROCURE PELOS QUADROS VERMELHOS FORA DA DIAGONAL")
    print("   └─→ Significa: 'Estas duas coisas andam JUNTAS'")
    print("   └─→ Exemplo: Se 'Pressão' e 'Temperatura' são vermelhas")
    print("   └─→ Significa: Quando pressão sobe, temperatura também sobe")
    
    print("\n3️⃣ PROCURE PELOS QUADROS AZUIS")
    print("   └─→ Significa: 'Estas duas coisas são OPOSTAS'")
    print("   └─→ Exemplo: Se 'Velocidade' e 'Tempo' são azuis")
    print("   └─→ Significa: Quanto mais rápido, menos tempo leva")
    
    print("\n4️⃣ PROCURE PELOS QUADROS BRANCOS")
    print("   └─→ Significa: 'Estas duas coisas são INDEPENDENTES'")
    print("   └─→ Exemplo: Se 'Cor' e 'Peso' são brancos")
    print("   └─→ Significa: Cor não afeta peso (andam separadas)")
    
    print("\n" + "="*80)
    print("🎯 O QUE ESTA MATRIZ NOS DIZ?")
    print("="*80)
    
    # Encontrar correlações fortes
    strong_corr = correlation_matrix[(correlation_matrix > 0.7) | (correlation_matrix < -0.7)].stack()
    strong_corr = strong_corr.drop_duplicates().sort_values(ascending=False)
    
    # Remover auto-correlação
    strong_corr_clean = strong_corr.drop([i for i in strong_corr.index if i[0] == i[1]])
    
    if len(strong_corr_clean) > 0:
        print("\n🚨 ENCONTRADAS CORRELAÇÕES FORTES (> 0.7 ou < -0.7):\n")
        for (var1, var2), valor in strong_corr_clean.head(10).items():
            if valor > 0.7:
                print(f"   ✓ {var1} ANDA COM {var2}")
                print(f"     │")
                print(f"     ├─ Valor: {valor:.2f}")
                print(f"     ├─ Tipo: CORRELAÇÃO POSITIVA ↗️")
                print(f"     └─ Significa: Quando {var1} aumenta,")
                print(f"                  {var2} também aumenta!\n")
            elif valor < -0.7:
                print(f"   ✗ {var1} ANDA OPOSTO A {var2}")
                print(f"     │")
                print(f"     ├─ Valor: {valor:.2f}")
                print(f"     ├─ Tipo: CORRELAÇÃO NEGATIVA ↙️")
                print(f"     └─ Significa: Quando {var1} aumenta,")
                print(f"                  {var2} diminui!\n")
    else:
        print("\n✨ NENHUMA CORRELAÇÃO FORTE ENCONTRADA!")
        print("\n💡 O QUE SIGNIFICA ISSO?")
        print("   ├─ Cada sensor funciona INDEPENDENTEMENTE")
        print("   ├─ Nenhum sensor 'puxar' o outro")
        print("   ├─ Isso é MUITO BOM! Significa:")
        print("   │  • Dados RICOS e DIVERSOS")
        print("   │  • Sem REDUNDÂNCIA (informação repetida)")
        print("   │  • Cada sensor traz informação NOVA")
        print("   └─ Nossos modelos de IA vão aprender MAIS!")

else:
    print("⚠️  Insuficientes variáveis numéricas para análise de correlação")
````

````python
# =======================================================================================
# 👶 EXPLICAÇÃO SUPER DIDÁTICA: ENTENDENDO CORRELAÇÃO COMO UMA CRIANÇA
# =======================================================================================
#
# Vamos usar EXEMPLOS DO DIA A DIA para entender correlação!
#
# ────────────────────────────────────────────────────────────────────────────────────
# ANALOGIA 1: MAÇÃ E FORÇA
# ────────────────────────────────────────────────────────────────────────────────────
#
# Imagine uma criança comendo maçã todo dia:
#
#     Dia 1: Come 1 maçã  → Tem força para brincar 1 hora
#     Dia 2: Come 2 maçãs → Tem força para brincar 2 horas
#     Dia 3: Come 3 maçãs → Tem força para brincar 3 horas
#
# RESULTADO: Correlação FORTE POSITIVA (+0.95)
# INTERPRETAÇÃO: Mais maçã = Mais força! 🍎💪
#
# ────────────────────────────────────────────────────────────────────────────────────
# ANALOGIA 2: CHUVA E ROUPA MOLHADA
# ────────────────────────────────────────────────────────────────────────────────────
#
# Observe uma criança brincando:
#
#     Dia SEM chuva  → Roupa SECA ✓
#     Dia COM chuva  → Roupa MOLHADA ✗
#     Dia SEM chuva  → Roupa SECA ✓
#
# RESULTADO: Correlação FORTE POSITIVA (+0.90)
# INTERPRETAÇÃO: Mais chuva = Mais roupa molhada! 🌧️👕
#
# ────────────────────────────────────────────────────────────────────────────────────
# ANALOGIA 3: VELOCIDADE E TEMPO DE CHEGADA
# ────────────────────────────────────────────────────────────────────────────────────
#
# Observe uma criança indo para a escola:
#
#     Velocidade LENTA (2 km/h)   → Tempo LONGO (1 hora)
#     Velocidade MÉDIA (4 km/h)   → Tempo MÉDIO (30 minutos)
#     Velocidade RÁPIDA (8 km/h)  → Tempo CURTO (15 minutos)
#
# RESULTADO: Correlação FORTE NEGATIVA (-0.95)
# INTERPRETAÇÃO: Quanto mais rápido, menos tempo! ⚡⏱️
#
# ────────────────────────────────────────────────────────────────────────────────────
# ANALOGIA 4: COR DO UNIFORME E NOTA DA PROVA
# ────────────────────────────────────────────────────────────────────────────────────
#
# Observe notas de crianças:
#
#     Uniforme AZUL   → Nota 7 (uma criança)
#     Uniforme AZUL   → Nota 9 (outra criança)
#     Uniforme AZUL   → Nota 5 (outra criança)
#     Uniforme VERDE  → Nota 8 (uma criança)
#     Uniforme VERDE  → Nota 6 (outra criança)
#
# RESULTADO: Correlação ZERO (0.05)
# INTERPRETAÇÃO: Cor do uniforme NÃO afeta a nota! 👕📝
#
# ────────────────────────────────────────────────────────────────────────────────────
# AGORA NA NOSSA ANÁLISE DE EQUIPAMENTOS:
# ────────────────────────────────────────────────────────────────────────────────────

print("\n" + "="*80)
print("🎓 VAMOS APLICAR ISTO AO PROGRAMA DESENVOLVER")
print("="*80)

print("\n❓ PERGUNTA 1: Pressão e Temperatura correlacionam?")
print("   └─ Se a resposta é SIM e CORRELAÇÃO FORTE POSITIVA:")
print("      → Quando pressão SOBE, temperatura também SOBE")
print("      → Podemos usar uma para PREVER a outra!")
print("      → Se pressão subir demais, já sabemos que temperatura")
print("        também vai subir (antes mesmo de medir)")

print("\n❓ PERGUNTA 2: Vibração e Tempo de Operação correlacionam?")
print("   └─ Se a resposta é SIM e CORRELAÇÃO FORTE POSITIVA:")
print("      → Quanto MAIS TEMPO o equipamento funciona,")
print("        MAIS vibração apareça")
print("      → Isto é NORMAL! Máquinas envelhecem e vibram mais")
print("      → Podemos PREVER quando vai quebrar!")

print("\n❓ PERGUNTA 3: Velocidade e Temperatura correlacionam?")
print("   └─ Se a resposta é NÃO (correlação próxima a 0):")
print("      → Velocidade alta NÃO significa temperatura alta")
print("      → Podem variar independentemente")
print("      → Um equipamento rápido pode ser frio")
print("      → Um equipamento lento pode ser quente")

print("\n" + "="*80)
print("🎯 CONCLUSÃO FINAL")
print("="*80)

print("\n💡 A Matriz de Correlação nos AJUDA A:")
print("   1. Identificar quais sensores trabalham JUNTOS")
print("   2. Entender relações CASUAIS nos equipamentos")
print("   3. Eliminar REDUNDÂNCIA (sensores que medem a mesma coisa)")
print("   4. Fazer PREVISÕES mais PRECISAS")
print("   5. ECONOMIZAR recursos (não precisar de todos os sensores)")

print("\n✨ PRONTO! Agora você entende CORRELAÇÃO como uma criança!")
print("   E isso é o primeiro passo para entender MACHINE LEARNING! 🚀")
print("="*80 + "\n")
````

## 🤖 Machine Learning - Modelagem Preditiva (RESPOSTA: P1)

### Objetivos desta Seção

✅ **Comparar performance** entre 5 modelos diferentes
✅ **Selecionar o melhor modelo** baseado em ROC-AUC
✅ **Gerar scores de risco** para equipamentos

### Por que múltiplos modelos?

Cada algoritmo tem **força e fraquezas**:
- **Regressão Logística**: Rápida, interpretável, baseline
- **Random Forest**: Robusto, capture interações entre features
- **Gradient Boosting**: Melhor generalização, reduz overfitting
- **SVM**: Eficaz em espaços altos, não-lineares
- **KNN**: Simples, local decision boundaries

**Comparar vários modelos aumenta confiança nas previsões!**

````python
# =======================================================================================
# TREINAMENTO DE 5 MODELOS DE MACHINE LEARNING
# =======================================================================================
# Processo:
#   1. Dividir dados em treino (80%) e teste (20%)
#   2. Treinar cada algoritmo no conjunto de treino
#   3. Avaliar no conjunto de teste
#   4. Comparar performance usando ROC-AUC
#
# Métrica Principal: ROC-AUC
#   - Varia de 0 a 1 (quanto maior, melhor)
#   - 0.5 = classificação aleatória
#   - 0.8+ = excelente modelo
# =======================================================================================

from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve
import warnings
warnings.filterwarnings('ignore')

print("🚀 TREINAMENTO DE MODELOS DE MACHINE LEARNING\n")

# STEP 1: Dividir dados com estratificação (mantém proporção de classes)
print("  ✓ Step 1: Dividindo dados em treino/teste (80/20) com estratificação")
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, 
    test_size=0.2,  # 20% para teste
    random_state=42,
    stratify=y  # Mantém proporção de classes
)

print(f"    - Treino: {X_train.shape[0]:,} amostras")
print(f"    - Teste: {X_test.shape[0]:,} amostras\n")

# STEP 2: Definir modelos
print("  ✓ Step 2: Configurando 5 algoritmos diferentes")
modelos = {
    'Regressão Logística': LogisticRegression(random_state=42, max_iter=1000),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    'Gradient Boosting': GradientBoostingClassifier(n_estimators=100, random_state=42),
    'SVM (RBF)': SVC(kernel='rbf', probability=True, random_state=42),
    'KNN (k=5)': KNeighborsClassifier(n_neighbors=5)
}

# STEP 3: Treinar e avaliar cada modelo
print("\n  ✓ Step 3: Treinando modelos...\n")

resultados_ml = {}
for nome, modelo in modelos.items():
    print(f"    📍 Treinando: {nome}")
    
    # Treinar no conjunto de treino
    modelo.fit(X_train, y_train)
    
    # Prever no conjunto de teste
    y_pred = modelo.predict(X_test)
    y_pred_proba = modelo.predict_proba(X_test)[:, 1]
    
    # Calcular ROC-AUC
    roc_auc = roc_auc_score(y_test, y_pred_proba)
    
    # Armazenar resultados
    resultados_ml[nome] = {
        'modelo': modelo,
        'y_pred': y_pred,
        'y_pred_proba': y_pred_proba,
        'roc_auc': roc_auc
    }
    
    print(f"       ROC-AUC: {roc_auc:.4f} ✓\n")

# STEP 4: Ranking dos modelos
print("\n" + "="*70)
print("🏆 RANKING FINAL DE MODELOS ML")
print("="*70)
for idx, (nome, resultado) in enumerate(
    sorted(resultados_ml.items(), key=lambda x: x[1]['roc_auc'], reverse=True), 1
):
    medal = ['🥇', '🥈', '🥉', '4️⃣', '5️⃣'][idx-1]
    print(f"{medal} {idx}. {nome:25s} → ROC-AUC = {resultado['roc_auc']:.4f}")
print("="*70)
````

### Vantagens vs. Machine Learning Tradicional

| Aspecto | Machine Learning | Deep Learning |
|--------|------------------|---------------|
| **Interpretabilidade** | Alta | Baixa (caixa preta) |
| **Dados Necessários** | Menos | Mais |
| **Padrões Complexos** | Limitado | Excelente |
| **Tempo Treino** | Rápido | Lento |
| **Hardware** | CPU | GPU recomendada |

### Arquiteturas Testadas

````python
# =======================================================================================
# CONFIGURAÇÃO DO AMBIENTE DE DEEP LEARNING
# =======================================================================================
# TensorFlow/Keras: Framework para Deep Learning
# GPU: Se disponível, acelera significativamente o treinamento
# =======================================================================================

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, Sequential
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

print("🧠 DEEP LEARNING - CONFIGURAÇÃO DO AMBIENTE\n")
print(f"  ✓ TensorFlow versão: {tf.__version__}")

gpu_devices = tf.config.list_physical_devices('GPU')
if gpu_devices:
    print(f"  ✓ GPU disponível: {gpu_devices}")
else:
    print(f"  ℹ️  GPU não disponível (usando CPU)")

# Preparar dados para Deep Learning
y_train_np = y_train.to_numpy() if hasattr(y_train, 'to_numpy') else y_train
y_test_np = y_test.to_numpy() if hasattr(y_test, 'to_numpy') else y_test

print(f"\n  ✓ Dados preparados para treino neural")
print(f"    - Shape de X_train: {X_train.shape}")
print(f"    - Shape de X_test: {X_test.shape}")

# =======================================================================================
# MODELO 1: REDE NEURAL SIMPLES
# =======================================================================================
# Arquitetura Simples:
#   - Input: N features
#   - Layer 1: 64 neurônios + ReLU + Dropout 30%
#   - Layer 2: 32 neurônios + ReLU + Dropout 20%
#   - Output: 1 neurônio + Sigmoid (probabilidade)
#
# Dropout: Técnica de regularização para evitar overfitting
# =======================================================================================

print("\n" + "="*70)
print("🔹 MODELO 1: REDE NEURAL SIMPLES")
print("="*70)

modelo_simples = Sequential([
    layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
    layers.Dropout(0.3),
    layers.Dense(32, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(1, activation='sigmoid')
])

modelo_simples.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy', keras.metrics.AUC()]
)

print("Arquitetura do modelo:")
modelo_simples.summary()

# Callback para parar treinamento se não melhorar
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

print("\n⏳ Treinando Modelo 1...")
historico_simples = modelo_simples.fit(
    X_train, y_train_np,
    epochs=50,
    batch_size=32,
    validation_split=0.2,  # 20% dos dados de treino para validação
    callbacks=[early_stop],
    verbose=0
)

print(f"✓ Treinamento concluído em {len(historico_simples.history['loss'])} épocas")

# Avaliação
y_pred_simples = modelo_simples.predict(X_test, verbose=0)
auc_simples = roc_auc_score(y_test_np, y_pred_simples)
print(f"  ROC-AUC Score (Teste): {auc_simples:.4f}")

print("\n" + "="*70)
print("🔹 MODELO 2: REDE NEURAL PROFUNDA")
print("="*70)

# =======================================================================================
# MODELO 2: REDE NEURAL PROFUNDA
# =======================================================================================
# Arquitetura Profunda:
#   - Input: N features
#   - Layer 1: 128 neurônios + ReLU + BatchNorm + Dropout 40%
#   - Layer 2: 64 neurônios + ReLU + BatchNorm + Dropout 30%
#   - Layer 3: 32 neurônios + ReLU + Dropout 20%
#   - Layer 4: 16 neurônios + ReLU + Dropout 10%
#   - Output: 1 neurônio + Sigmoid
#
# BatchNormalization: Normaliza ativações entre layers, acelera convergência
# =======================================================================================

modelo_profundo = Sequential([
    layers.Dense(128, activation='relu', input_shape=(X_train.shape[1],)),
    layers.BatchNormalization(),
    layers.Dropout(0.4),
    
    layers.Dense(64, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.3),
    
    layers.Dense(32, activation='relu'),
    layers.Dropout(0.2),
    
    layers.Dense(16, activation='relu'),
    layers.Dropout(0.1),
    
    layers.Dense(1, activation='sigmoid')
])

modelo_profundo.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss='binary_crossentropy',
    metrics=['accuracy', keras.metrics.AUC()]
)

print("Arquitetura do modelo:")
modelo_profundo.summary()

print("\n⏳ Treinando Modelo 2...")
historico_profundo = modelo_profundo.fit(
    X_train, y_train_np,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    callbacks=[early_stop],
    verbose=0
)

print(f"✓ Treinamento concluído em {len(historico_profundo.history['loss'])} épocas")

# Avaliação
y_pred_profundo = modelo_profundo.predict(X_test, verbose=0)
auc_profundo = roc_auc_score(y_test_np, y_pred_profundo)
print(f"  ROC-AUC Score (Teste): {auc_profundo:.4f}")

print("\n" + "="*70)
````

````python
# Modelo 2: Rede Neural Profunda
print("\n" + "="*50)
print("Modelo 2: Rede Neural Profunda (5 camadas)")
print("="*50)

modelo_profundo = Sequential([
    layers.Dense(128, activation='relu', input_shape=(X_train.shape[1],)),
    layers.BatchNormalization(),
    layers.Dropout(0.4),
    
    layers.Dense(64, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.3),
    
    layers.Dense(32, activation='relu'),
    layers.Dropout(0.2),
    
    layers.Dense(16, activation='relu'),
    layers.Dropout(0.1),
    
    layers.Dense(1, activation='sigmoid')
])

modelo_profundo.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss='binary_crossentropy',
    metrics=['accuracy', keras.metrics.AUC()]
)

print(modelo_profundo.summary())

historico_profundo = modelo_profundo.fit(
    X_train, y_train_np,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    callbacks=[early_stop],
    verbose=0
)

print(f"Épocas treinadas: {len(historico_profundo.history['loss'])}")

# Avaliação
y_pred_profundo = modelo_profundo.predict(X_test, verbose=0)
auc_profundo = roc_auc_score(y_test_np, y_pred_profundo)
print(f"ROC-AUC Score (Teste): {auc_profundo:.4f}")
````
